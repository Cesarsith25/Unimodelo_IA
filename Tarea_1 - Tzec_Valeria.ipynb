## Tuplas 'tuple'

**Definición**

Sirven para almacenar múltiples items en una variable. Es una colección ordenada e inmutable (sus elementos no se pueden modificar, agregar ni eliminar una vez creada). Una tupla puede tener diferentes tipos de datos).

**Sintaxis básica**

```python
tuplaEjemplo1 = ("abc", 34, True)
```

```python
tuplaEjemplo2 = ("hola",) #Cuando la tupla solo contiene 1 elemento, 
#este lleva coma al final
```

**Uso en ciclos**

```python
for e in tuplaEjemplo1:
    print(e)
```

**Métodos de uso frecuente**

`.count(x)` -> Devuelve el número de veces que aparece el valor `x`

`.index(x)` -> Devuelve la primera posición en la que aparece el valor `x`

## Conjuntos 'set'

**Definición**

Sirven para almacenar múltiples items en una variable. Colección desordenada de elementos únicos y mutable. Los elementos son inmutables, sin embargo, pueden ser removidos y nuevos elementos ser añadidos.

**Sintaxis básica**

```python
conjuntoEjemplo1 = {1, 2, 3, 4, True, "hola", 4}
```

```python
conjunto_vacio = set() #{} crea un diccionario vacío, por lo que se usa set()
```

**Uso en ciclos**

```python
numeros = {1, 4, 2}
```

```python
for num in numeros:
    print(num)
```

**Métodos de uso frecuente**

`.len(thisset)`: Retorna la cantidad de elementos del conjunto.

`.add(x)`: Añade el elemento `x` al conjunto.

`.remove(x)` / `.discard(x)`: Elimina el elemento `x` (`remove` genera un error si no existe; `discard` no).

`.union(otro)`: Retorna un conjunto con la unión de ambos conjuntos (operador `|`).

`.intersection(otro)`: Retorna los elementos compartidos (operador `&`).

`.difference(otro)`: Retorna los elementos presentes únicamente en el conjunto principal (operador).

## Diccionarios 'dict'

**Definición**

Colección basada en pares clave:valor. Mantienen el orden de inserción (desde Python 3.7, versiones anteriores eran desordenados). Las claves deben ser únicas e inmutables, mientras que los valores pueden ser de cualquier tipo. Se escriben con corchetes.

**Sintaxis básica**

```python
usuario = {
    "nombre": "Vale",
    "edad": 20,
    "activo": False,
    "activo": True
}
```

**Uso en ciclos**

```python
#Iterar clave y valor al mismo tiempo
```

```python
for clave, valor in usuario.items():
    print(f"{clave}:{valor}")
```

```python
#Iterar solo por claves o solo por valores
```

```python
for clave in usuario.keys():
    print(f"{clave}")
```

```python
for valor in usuario.values():
    print(f"{valor}")
```

**Métodos de uso frecuente**

`.get(clave, default)`: Retorna el valor de la clave indicada sin lanzar error si esta no existe.

`.keys()` / `.values()` / `.items()`: Retornan vistas de las claves, los valores o los pares clave-valor.

`.pop(clave)`: Elimina la clave indicada y retorna su valor asociado.

`.update()otro_dict`: Actualiza el diccionario insertando o modificando claves de otro diccionario

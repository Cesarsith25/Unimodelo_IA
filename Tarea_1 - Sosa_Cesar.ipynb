# Estructuras de Datos en Python

## 1. Tuplas (`tuple`)

### Definición
Sirven para almacenar múltiples ítems en una sola variable. Es una colección **ordenada** e **inmutable** (sus elementos no se pueden modificar, agregar ni eliminar una vez creada). Una tupla puede contener diferentes tipos de datos.

### Sintaxis básica y uso de ciclos

```python
# Creación de tuplas con distintos tipos de datos
tupla_ejemplo1 = ("abc", 34, True, 34)
tupla_ejemplo1
```

```python
# Cuando la tupla solo contiene 1 elemento, este lleva coma al final
tupla_ejemplo2 = ("hola",)
tupla_ejemplo2
```

### Métodos frecuentes de Tuplas
* `.count(x)`: Devuelve el número de veces que aparece el valor `x` dentro de la tupla.
* `.index(x)`: Devuelve el índice de la primera posición donde aparece el valor `x`.

```python
productos = ("huevos", "leche", "harina", "aceite", "huevos")
```

```python
# Contar cuántas veces se repite un elemento
cantidad_huevos = productos.count("huevos")
print("Cantidad de huevos:", cantidad_huevos)
```

```python
# Obtener la primera posición de un elemento
posicion_leche = productos.index("leche")
print("Índice de leche:", posicion_leche)
```

## 2. Conjuntos (`set`)

### Definición
Sirven para almacenar múltiples ítems en una sola variable. Es una colección **desordenada** de elementos **únicos** y **mutable** (los elementos internos son inmutables, pero se pueden añadir o remover elementos del conjunto).

### Sintaxis básica y uso de ciclos

```python
# Creación de conjuntos (los elementos duplicados se eliminan automáticamente)
conjunto_ejemplo1 = {1, 2, 3, 4, True, "hola", 4}
conjunto_ejemplo1
```

```python
# Para crear un conjunto vacío se usa set()
conjunto_vacio = set()
conjunto_vacio
```

```python
# Recorrer un conjunto con un ciclo for (el orden de salida puede variar)
for num in conjunto_ejemplo1:
    print(num)
```

### Métodos frecuentes de Conjuntos
* `len(set)`: Retorna la cantidad total de elementos únicos en el conjunto.
* `.add(x)`: Añade el elemento `x` al conjunto.
* `.remove(x)` / `.discard(x)`: Eliminan el elemento `x`. `.remove()` lanza un error `KeyError` si el elemento no existe; `.discard()` no genera error.
* `.union(otro)`: Retorna un nuevo conjunto con todos los elementos de ambos (operador `|`).
* `.intersection(otro)`: Retorna solo los elementos que comparten ambos conjuntos (operador `&`).
* `.difference(otro)`: Retorna los elementos presentes únicamente en el conjunto principal (operador `-`).

```python
conjunto_a = {1, 2, 3, 4}
conjunto_b = {3, 4, 5, 6}
```

```python
# 1. Cantidad de elementos
print("Longitud de A:", len(conjunto_a))
```

```python
# 2. Agregar elementos
conjunto_a.add(10)
print("A tras add(10):", conjunto_a)
```

```python
# 3. Eliminar elementos
conjunto_a.remove(10)   # Elimina 10 (lanza error si 10 no existe)
print(conjunto_a)
conjunto_a.discard(99)  # Intenta eliminar 99 (no lanza error si no existe)
print(conjunto_a)
```

```python
# 4. Operaciones entre conjuntos
union_res = conjunto_a.union(conjunto_b)             # O bien: conjunto_a | conjunto_b
print(union_res)
```

```python
intersec_res = conjunto_a.intersection(conjunto_b)   # O bien: conjunto_a & conjunto_b
print(intersec_res)
```

```python
diferencia_res = conjunto_a.difference(conjunto_b)   # O bien: conjunto_a - conjunto_b
print(diferencia_res)
```

## 3. Diccionarios (`dict`)

### Definición
Colección de elementos almacenados en pares `clave: valor`. 
* Mantienen el **orden de inserción** (a partir de Python 3.7+).
* Las **claves** deben ser **únicas** e **inmutables** (strings, números, tuplas).
* Los **valores** pueden ser de cualquier tipo de dato y pueden repetirse.

### Sintaxis básica y uso de ciclos

```python
# Creación de un diccionario
usuario = {
    "nombre": "Vale",
    "edad": 20,
    "activo": False,
    "activo": True  # Las claves duplicadas sobrescriben el valor anterior
}
usuario
```

```python
# 1. Iterar clave y valor al mismo tiempo con .items()
print("--- Clave y Valor ---")
for clave, valor in usuario.items():
    print(f"{clave}: {valor}")
```

```python
# 2. Iterar solo por claves con .keys()
print("--- Claves ---")
for clave in usuario.keys():
    print(clave)
```

```python
# 3. Iterar solo por valores con .values()
print("--- Valores ---")
for valor in usuario.values():
    print(valor)
```

### Métodos frecuentes de Diccionarios
* `.get(clave, default)`: Retorna el valor de la clave indicada. Si la clave no existe, devuelve `None` (o el valor `default` especificado) en lugar de lanzar un error `KeyError`.
* `.keys()` / `.values()` / `.items()`: Retornan vistas dinámicas de las claves, los valores o los pares clave-valor.
* `.pop(clave)`: Elimina la clave indicada del diccionario y retorna su valor asociado.
* `.update(otro_dict)`: Actualiza el diccionario insertando o modificando pares clave-valor provenientes de otro diccionario.

```python
perfil = {
    "usuario": "dev_player",
    "nivel": 45,
    "rol": "backend"
}
```

```python
# 1. Uso de .get() seguro (evita errores si la clave no existe)
rango = perfil.get("rango", "Sin Rango")  # Clave inexistente: retorna "Sin Rango"
print("Rango:", rango)
```

```python
# 2. Uso de .pop() para eliminar y extraer un valor
nivel_extraido = perfil.pop("nivel")
print("Nivel eliminado:", nivel_extraido)
print("Perfil tras pop():", perfil)
```

```python
# 3. Uso de .update() para agregar o actualizar pares clave-valor
perfil.update({
    "rol": "fullstack",   # Modifica la clave existente "rol"
    "lenguaje": "Python" # Añade una nueva clave "lenguaje"
})
```

```python
print("Perfil actualizado:", perfil)
```

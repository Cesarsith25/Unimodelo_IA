```python
import numpy as np
import matplotlib.pyplot as plt
```

<h2> Example with $h_\theta(x) = \theta\cdot x = \theta_1 x_1$</h2>

```python
m = 6

X = np.linspace(0, 5, m)
y = X

print (X)
print (y)
```

```python
plt.figure (figsize = (5, 5))

theta1 = 0.5
h = theta1 * X

plt.plot(X, y, 'ro', markersize = 6)
plt.plot(X, h, 'b-', linewidth = 2)

plt.xlabel('X', fontsize = 12)
plt.xlabel('y', fontsize = 12)
plt.grid(True)
plt.legend(['y', 'h'])
plt.show()
```

```python
theta1 = np.linspace(-3, 5, 101)
h = [theta * X for theta in theta1]
J = (1 / m) * (np.abs(h - y)).sum(axis = 1) #MAE

plt.plot(theta1, J, 'b-', linewidth = 2)

plt.xlabel(r'$\theta_1$', fontsize = 12)
plt.ylabel(r'$J(\theta_1)$', fontsize = 12)
plt.grid(True)
plt.show()
```

<h3>MSE Nos sirve este para entrenar, sin embargo, al arrojar un resultado cuadrado, las unidades no cuadran para mostrar el error</h3>

```python
theta1 = np.linspace(-3, 5, 101)
h = [theta * X for theta in theta1]
J = (1 / m) * (np.square(h - y)).sum(axis = 1) #MSE

plt.plot(theta1, J, 'b-', linewidth = 2)

plt.xlabel(r'$\theta_1$', fontsize = 12)
plt.ylabel(r'$J(\theta_1)$', fontsize = 12)
plt.grid(True)
plt.show()
```

```python
theta1 = np.linspace(-3, 5, 101)
h = [theta * X for theta in theta1]
J = np.sqrt((1 / m) * (np.square(h - y)).sum(axis = 1)) #RMSE

plt.plot(theta1, J, 'b-', linewidth = 2)

plt.xlabel(r'$\theta_1$', fontsize = 12)
plt.ylabel(r'$J(\theta_1)$', fontsize = 12)
plt.grid(True)
plt.show()
```

<h2> Example with $h_\theta(x) = \theta_0 + \theta_1 x_1$</h2>

```python
m = 10

X = np.linspace(-2, 2, m)
y = 2.5 + 5 * X
```

```python
theta0 = np.linspace(-2.5, 7.5, 50)
theta1 = np.linspace(0, 10, 50)

h = np.array([[th0 + th1 * X for th0 in theta0] for th1 in theta1])
J = (1 / m) * ((h - y)**2).sum(axis = 2) #MSE

fig = plt.figure()
ax = plt.axes(projection = '3d')
th0, th1 = np.meshgrid(theta0, theta1)
surf = ax.plot_surface(th0, th1, J, cmap = 'coolwarm')

ax.set_xlabel(r'$\theta_0$', fontsize = 12)
ax.set_ylabel(r'$J(\theta_1)$', fontsize = 12)
ax.set_zlabel(r'$J(\theta_1, \theta_1)$', fontsize = 12)

fig.colorbar(surf, shrink = 0.5, aspect = 5)
ax.view_init(20, 45)
plt.show()
```

```python
plt.contour(th0, th1, J, 50, cmap = 'coolwarm')
ax.set_xlabel(r'$\theta_0$', fontsize = 12)
ax.set_ylabel(r'$\theta_1$', fontsize = 12)
plt.colorbar()
plt.show()
```

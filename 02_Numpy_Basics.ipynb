```python
A = np.array([[1, 2, 0, 3]])
B = np.array([[-1, 1, 2, 1]])
```

```python
print(A.shape)
print(B.shape)
```

```python
np.dot(A, B) # no cumple con (m,k),(k,n)
```

```python
# matriz.T -> T hace referencia a transpuesta
```

```python
np.dot(A, B.T) # No recomendado
```

```python
np.matmul(A, B.T)
```

```python
A @ B.T # @ hace lo mismo que matmul
```

```python
A = np.array([[1, 2, 0, 3], [-1, 2, 3, 0]])
B = np.array([[-1, 1, 2], [1, 2, 1], [0, -1, -2], [1, 1, 1]])
```

```python
print(A.shape)
print(B.shape)
```

```python
# (2,4) x (4,3) => (2,3)
A @ B
```

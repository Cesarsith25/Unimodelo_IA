```python
def fx(*args):
    print(type(args))
    print(args)
    print(len(args))
    for e in args:
        print(e)
```

```python
fx('hola', 'mundo', 3.56, True)
```

```python
def fx2(**kwargs):
    print(type(kwargs))
    print(kwargs)
    print(len(kwargs))
    for e in kwargs:
        print(e)
```

```python
fx2(nombre='Pepe', asegurado=True, peso=86, altura=1.75)
```
